```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([2, 4, 6, 8, 10, 12]).reshape(-1, 1)
y = np.array([3.0, 5.5, 7.2, 9.5, 12.0, 14.5])
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
new_screens = np.array([[7]])
predicted_cost = model.predict(new_screens)

print(f"Прогнозована вартість додатку на {new_screens[0][0]} екранів: {predicted_cost[0]:.2f} тис. $")
```

```python
y_pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', s=50, label='Історичні дані (Фактичні замовлення)')

plt.plot(X, y_pred, color='red', linewidth=2, label='Модель (Лінійна регресія)')

plt.scatter(new_screens, predicted_cost, color='green', s=150, marker='*', zorder=5, label='Прогноз для 7 екранів')

plt.title('Прогнозування вартості розробки мобільного додатку')
plt.xlabel('Кількість екранів (шт.)')
plt.ylabel('Вартість (тис. $)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
